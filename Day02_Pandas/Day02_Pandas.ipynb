# Day 02 — Python & Pandas Fundamentals

Practise Pandas operations using the Student Performance, Nigeria House Prices, and Stroke Disease datasets.

**Before running:** Update the three CSV paths below. Student Performance is semicolon-separated; the other datasets are comma-separated.

```python
import pandas as pd

# Update these paths to the locations of your CSV files
# Change these paths if your files are stored elsewhere.

student_path = "student-por.csv"

houses_path = "nigeria_houses_data.csv"

stroke_path = "StrokeDisease.csv"

student_df = pd.read_csv(student_path, sep=";")
houses_df = pd.read_csv(houses_path)
stroke_df = pd.read_csv(stroke_path)

print("All three datasets loaded successfully.")
```
