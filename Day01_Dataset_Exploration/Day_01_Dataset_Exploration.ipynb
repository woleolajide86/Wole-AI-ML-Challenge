```python

# ============================================================
# DAY 1: DATASET EXPLORATION
# 30-DAY AI/ML CHALLENGE
# ============================================================

# SECTION 1: IMPORT LIBRARIES

import pandas as pd

print("=" * 65)
print("DAY 1: DATASET EXPLORATION")
print("=" * 65)
```

```python
# ============================================================
# SECTION 2: DATASET FILE PATHS
# ============================================================

# Change these paths if your files are stored elsewhere.

student_path = "student-por.csv"

houses_path = "nigeria_houses_data.csv"

stroke_path = "StrokeDisease.csv"
```

```python

# ============================================================
# SECTION 3: LOAD THE DATASETS
# ============================================================

# Student dataset uses semicolon (;)
student_df = pd.read_csv(student_path, sep=";")

# House dataset uses comma (,)
houses_df = pd.read_csv(houses_path)

# Stroke dataset uses comma (,)
stroke_df = pd.read_csv(stroke_path)
```

```python

# ============================================================
# SECTION 4: STORE DATASETS IN ONE DICTIONARY
# ============================================================

datasets = {

    "Student Performance": {
        "data": student_df,
        "target": "G3",
        "problem": "Regression"
    },

    "Nigeria House Prices": {
        "data": houses_df,
        "target": "price",
        "problem": "Regression"
    },

    "Stroke Disease": {
        "data": stroke_df,
        "target": "stroke",
        "problem": "Binary Classification"
    }

}

```

```python

# ============================================================
# SECTION 5A: STUDENT PERFORMANCE DATASET EXPLORATION
# ============================================================

# Select the Student Performance Dataset
df = student_df

print("=" * 65)
print("STUDENT PERFORMANCE DATASET")
print("=" * 65)


# 1. NUMBER OF OBSERVATIONS (ROWS)

print("\nNUMBER OF OBSERVATIONS:")
print(df.shape[0])


# 2. NUMBER OF COLUMNS

print("\nNUMBER OF COLUMNS:")
print(df.shape[1])


# 3. COLUMN NAMES

print("\nCOLUMN NAMES:")

for column in df.columns:
    print("-", column)


# 4. FIRST FIVE ROWS

print("\nFIRST FIVE ROWS:")
print(df.head())


# 5. LAST FIVE ROWS

print("\nLAST FIVE ROWS:")
print(df.tail())


# 6. DATA TYPES

print("\nDATA TYPES:")
print(df.dtypes)


# 7. NUMERICAL VARIABLES

numerical = df.select_dtypes(
    include=["number"]
).columns.tolist()

print("\nNUMERICAL VARIABLES:")
print(numerical)


# 8. CATEGORICAL VARIABLES

categorical = df.select_dtypes(
    exclude=["number"]
).columns.tolist()

print("\nCATEGORICAL VARIABLES:")
print(categorical)


# 9. TARGET VARIABLE

target = "G3"

print("\nTARGET VARIABLE:")
print(target)


# 10. MACHINE LEARNING PROBLEM

print("\nSUITABLE MACHINE LEARNING PROBLEM:")
print("Regression")


# 11. IDENTIFY INPUT FEATURES

features = [
    column for column in df.columns
    if column != target
]

print("\nPREDICTIVE INPUT FEATURES:")
print(features)

print("\nNUMBER OF PREDICTIVE FEATURES:")
print(len(features))


print("\nSTUDENT PERFORMANCE EXPLORATION COMPLETED!")

```

```python

# ============================================================
# SECTION 5C: STROKE DISEASE DATASET EXPLORATION
# ============================================================

# Select the Stroke Disease Dataset
df = stroke_df

print("=" * 65)
print("STROKE DISEASE DATASET")
print("=" * 65)


# 1. NUMBER OF OBSERVATIONS (ROWS)

print("\nNUMBER OF OBSERVATIONS:")
print(df.shape[0])


# 2. NUMBER OF COLUMNS

print("\nNUMBER OF COLUMNS:")
print(df.shape[1])


# 3. COLUMN NAMES

print("\nCOLUMN NAMES:")

for column in df.columns:
    print("-", column)


# 4. FIRST FIVE ROWS

print("\nFIRST FIVE ROWS:")
print(df.head())


# 5. LAST FIVE ROWS

print("\nLAST FIVE ROWS:")
print(df.tail())


# 6. DATA TYPES

print("\nDATA TYPES:")
print(df.dtypes)


# 7. NUMERICAL VARIABLES

numerical = df.select_dtypes(
    include=["number"]
).columns.tolist()

print("\nNUMERICAL VARIABLES:")
print(numerical)


# 8. CATEGORICAL VARIABLES

categorical = df.select_dtypes(
    exclude=["number"]
).columns.tolist()

print("\nCATEGORICAL VARIABLES:")
print(categorical)


# 9. TARGET VARIABLE

target = "stroke"

print("\nTARGET VARIABLE:")
print(target)


# 10. MACHINE LEARNING PROBLEM

print("\nSUITABLE MACHINE LEARNING PROBLEM:")
print("Binary Classification")


# 11. IDENTIFY INPUT FEATURES

# Exclude the target variable and ID column.
# ID is an identifier, not a predictive feature.

features = [
    column for column in df.columns
    if column not in ["stroke", "id"]
]

print("\nPREDICTIVE INPUT FEATURES:")
print(features)

print("\nNUMBER OF PREDICTIVE FEATURES:")
print(len(features))


print("\nSTROKE DISEASE EXPLORATION COMPLETED!")
```

```python


# ============================================================
# SECTION 6: FINAL DATASET SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("FINAL DATASET SUMMARY")
print("=" * 80)

# Create an empty list to store the summary of each dataset
summary_data = []

# Loop through all datasets
for dataset_name, information in datasets.items():

    df = information["data"]

    target = information["target"]

    # Identify predictive features
    features = [
        column for column in df.columns
        if column != target
    ]

    # Exclude the ID column from Stroke Disease predictors
    if dataset_name == "Stroke Disease":
        features.remove("id")

    # Store the dataset information
    summary_data.append({
        "Dataset": dataset_name,
        "Observations": len(df),
        "Total Columns": len(df.columns),
        "Predictive Features": len(features),
        "Target Variable": target,
        "ML Problem": information["problem"]
    })


# Convert the summary into a Pandas DataFrame
summary_table = pd.DataFrame(summary_data)

# Display the summary table
print("\n")
print(summary_table.to_string(index=False))

print("\n")
print("=" * 80)
print("DAY 1 DATASET EXPLORATION COMPLETED!")
print("=" * 80)

# ============================================================
# END OF DAY 1
# ============================================================
```
